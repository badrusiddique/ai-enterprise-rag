```python
# Import the packages that register and use the local spaCy transformer model.
import nltk
import spacy
import spacy_transformers
import spacy_curated_transformers
import spacy_curated_transformers.models

from guardrails import Guard
from guardrails_ai.competitor_check import CompetitorCheck

from dotenv import load_dotenv
```

```python
# Load OPENAI_API_KEY and other local settings from the project .env file.
load_dotenv()
```

```python
# Create a validator that rejects text mentioning a competitor.
competitor_names = ["Apple", "Google", "Microsoft", "Samsung", "Amazon"]

def create_competitor_check_guard(on_fail):
    competitor_validator = CompetitorCheck(
        competitors=competitor_names,
        on_fail=on_fail,
    )
    return Guard().use(competitor_validator)
```

```python
# Compare a statement that mentions a competitor with a statement that does not.
competitor_statement = "Apple is a leading technology company known for its innovative products."
general_statement = "I like all the fruits, but apple and grapes are my favorite."
```

```python
# Add the validator to a Guard for easy input validation.
guard = create_competitor_check_guard(on_fail="exception")

# The competitor statement should fail validation.
try:
    competitor_result = guard.validate(competitor_statement)
    print("Unexpected success:", competitor_result)
except Exception as validation_error:
    print("Competitor statement rejected:", validation_error)

# The general statement should pass because "apple" is used as a fruit.
try:
    general_result = guard.validate(general_statement)
    print("General statement accepted:", general_result)
except Exception as validation_error:
    print("Unexpected validation failure:", validation_error)
```

```python
# Add the validator to a Guard for easy input validation.
guard = create_competitor_check_guard(on_fail="reask")

# The competitor statement should fail validation.
try:
    competitor_result = guard.validate(competitor_statement)
    print("Competitor statement with reask accepted:", competitor_result)
    print("Unexpected success:", competitor_result)
except Exception as validation_error:
    print("Unexpected success:", validation_error)

# The general statement should pass because "apple" is used as a fruit.
try:
    general_result = guard.validate(general_statement)
    print("General statement accepted:", general_result)
except Exception as validation_error:
    print("Unexpected validation failure:", validation_error)
```

```python
# Add the validator to a Guard for easy input validation.
guard = create_competitor_check_guard(on_fail="fix")

# The competitor statement should fail validation.
try:
    competitor_result = guard.validate(competitor_statement)
    print("Competitor statement with fix accepted:", competitor_result)
    print("Unexpected success:", competitor_result)
except Exception as validation_error:
    print("Unexpected success:", validation_error)

# The general statement should pass because "apple" is used as a fruit.
try:
    general_result = guard.validate(general_statement)
    print("General statement accepted:", general_result)
except Exception as validation_error:
    print("Unexpected validation failure:", validation_error)
```
